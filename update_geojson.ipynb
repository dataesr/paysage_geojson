```python
import pandas as pd

from config import FICHIER_CARTE, FICHIER_CARTE_UPDATE
from fonctions_geo import get_uu, get_dep, get_aca, get_reg, get_pays, ids_manquants
from fonctions_export import update_carte, save_dict, dump_mongo
```

```python
df = pd.read_json(FICHIER_CARTE, dtype={'originalId': str})
df_all = df.copy()
df.level.value_counts()
```

## Unités urbaines

```python
dict_uu = get_uu()
ids_manquants(df, 'urbanUnity', dict_uu)
```

## Départements

```python
dict_dep = get_dep()
ids_manquants(df, 'department', dict_dep)
```

## Académies
Union des départements de chaque académie

```python
dict_aca = get_aca(df, dict_dep)
ids_manquants(df, 'academy', dict_aca)
```

## Régions
Inclut R00 = regroupement des collectivités d'outre-mer

```python
dict_reg = get_reg()
ids_manquants(df, 'region', dict_reg)
```

## Pays

```python
dict_world = get_pays(df)
ids_manquants(df, 'country', dict_world)
```

## Remplacement des géométries

```python
save_dict(dict_aca, 'aca.json')
save_dict(dict_dep, 'dep.json')
save_dict(dict_reg, 'region.json')
save_dict(dict_world, 'pays.json')
```

```python
dict_all = dict_world | dict_uu | dict_dep | dict_aca | dict_reg
# dict_all = dict_uu   # pour ne traiter qu'un seul niveau
```

```python
df_all = update_carte(df_all, dict_all)
df_all.to_json(FICHIER_CARTE_UPDATE, orient='records')
```

## Exporter dans MongoDB

```python
dump_mongo(FICHIER_CARTE_UPDATE)
```
